First want to load our data

```python
from pathlib import Path
from mth5.mth5 import MTH5

data_path = Path.home() / "magnetotellurics" / "projects" / "resistics_mth5"
mth5_object = MTH5()
mth5_object.open_mth5(data_path / "8P_CAS04_NVR08.h5")
mth5_object
```

```python
cas04 = mth5_object.get_station("CAS04", survey="CONUS_South")
run_a = cas04.get_run("a")
chan = run_a.get_channel("hy")
ts = chan.to_channel_ts()
print(ts)
ts.plot()
```
